```python
co2_data = pd.read_csv("../data/raw/owid-co2-data.csv")
```

```python
co2_data.shape
```

```python
co2_data.head()
```

```python
energy_data = data[
    [
        'country',
        'year',
        'renewables_share_energy'
    ]
]
```

```python
co2_clean = co2_data[
    [
        'country',
        'year',
        'co2'
    ]
]
```

```python
merged_data = pd.merge(
    energy_data,
    co2_clean,
    on=['country', 'year'],
    how='inner'
)
```

```python
merged_data.shape
```

```python
merged_data.head()
```

```python
countries = [
    "Germany",
    "India",
    "China",
    "United States",
    "France",
    "Spain",
    "Italy"
]

project_data = merged_data[
    merged_data['country'].isin(countries)
]
```

```python
project_data = project_data[
    (project_data['year'] >= 2000)
    &
    (project_data['year'] <= 2023)
]
```

```python
project_data = project_data.dropna()
```
